### 1. Вывести все доступные платформы, на которых выпускалась хотя-бы одна игра;

```python
import pandas as pd

vgsales = pd.read_csv('vgsales.csv')

platforms = vgsales['Platform'].unique()
print(f"Всего платформ: {len(platforms)}\n")
print("Список платформ:")
print(sorted(platforms))
```

### 2. Добавить к копии набора данных столбец metacritic_rating и заполнить его данными из набора данных <a href='https://www.kaggle.com/skateddu/metacritic-all-time-games-stats'>Metacritic all time games stats</a>;


```python
metacritic = pd.read_csv('metacritic_games.csv')
vgsales_copy = vgsales.copy()

metacritic_clean = metacritic[['name', 'platform', 'rating']].copy()
metacritic_clean['name_lower'] = metacritic_clean['name'].str.lower().str.strip()
metacritic_clean['platform_lower'] = metacritic_clean['platform'].str.upper().str.strip()

vgsales_copy['name_lower'] = vgsales_copy['Name'].str.lower().str.strip()
vgsales_copy['platform_lower'] = vgsales_copy['Platform'].str.upper().str.strip()

vgsales_merged = vgsales_copy.merge(
    metacritic_clean[['name_lower', 'rating']],
    on='name_lower',
    how='left'
)

vgsales_merged = vgsales_merged.rename(columns={'rating': 'metacritic_rating'})
vgsales_merged = vgsales_merged.drop(columns=['name_lower', 'platform_lower'])

display(vgsales_merged[['Name', 'Platform', 'Year', 'metacritic_rating']])
```

### 3. Вывести список игр, рейтинг которых равен "M" и год издания не ранее 2012 года;



```python
filtered_games = vgsales_merged[
    (vgsales_merged['metacritic_rating'] == 'M') & 
    (vgsales_merged['Year'].notna()) & 
    (vgsales_merged['Year'] >= 2012)
].copy()

display(filtered_games[['Name', 'Platform', 'Year', 'Genre', 'Publisher', 'Global_Sales', 'metacritic_rating']])
```

### 4. Рассчитать и вывести описательные статистики для списка, полученного в предыдущем пункте;

```python
print("Описательные статистики для числовых столбцов:\n")
display(filtered_games.describe())

print("\nСтатистики для категориальных столбцов:\n")
categorical_stats = filtered_games[['Name', 'Platform', 'Genre', 'Publisher', 'metacritic_rating']].describe()
display(categorical_stats)
```

### 5. Вывести жанры игр с количеством игр, которые принадлежат соответствующему жанру в виде "<жанр> - <количество игр>" для всех жанров, которые содержат в своем названии не менее 3 различных гласных.

```python
def count_vowels(text):
    """Подсчитывает количество уникальных гласных букв в строке"""
    if pd.isna(text):
        return 0
    vowels = set('aeiouAEIOU')
    return len(set(char for char in str(text) if char in vowels))

genre_counts = vgsales.groupby('Genre').size().reset_index(name='game_count')

genre_counts['vowel_count'] = genre_counts['Genre'].apply(count_vowels)

filtered_genres = genre_counts[genre_counts['vowel_count'] >= 3].copy()

filtered_genres = filtered_genres.sort_values('game_count', ascending=False)

print(f"Жанры с ≥3 различными гласными в названии = {len(filtered_genres)}:\n")
print("<жанр> - <количество игр>\n")

for _, row in filtered_genres.iterrows():
    print(f"{row['Genre']} - {row['game_count']}")

print("Детализация жанр | гласные | кол-во игр:")
for _, row in filtered_genres.iterrows():
    vowels_in_genre = set(c.lower() for c in str(row['Genre']) if c.lower() in 'aeiou')
    print(f"{row['Genre']:20} | {sorted(vowels_in_genre)} | {row['game_count']}")
```
