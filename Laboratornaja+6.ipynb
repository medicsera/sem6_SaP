 > ## EX-1-5
 >
 > Используя оператор print и его именованные аргументы, выведите значения целых переменных x=5 и y=3 в следующих форматах: 
 >- 3+5=?
 >- 3 | 5 | 8
 >- Z(5)=F(3)
 >- x=5; y=3;
 >- Ответ: (5;3) 

```python
x = 5
y = 3
print(y, "+", x, "=", "?", sep="")
print(y, x, x+y, sep=" | ")
print("Z(", x, ")=", "F(", y, ")", sep="")
print("x=", x, "; y=", y, ";", sep="")
print("Ответ: ", "(", x, ";", y, ")", sep="")
```

 > ## EX-4-2
 >
 > Напишите программу, которая считывает целое неотрицательное число n и выводит его факториал. Факториал числа n обозначается n! и равен произведению всех натуральных чисел от 1 до  n. 

```python
n = int(input())
fact = 1
for i in range(1, n + 1):
    fact *= i
print(fact)
```

 > ## EX-4-4
 >
 > Выведите пирамиду из символов « » заданной высоты. Пирамиду, не прямоугольный треугольник :) 

```python
h = int(input())
for i in range(1, h + 1):
    print(" " * (h - i) + "*" * (2 * i - 1))
```

 > ## EX-4-5
 >
 > Напишите программу построения прямоугольника размерами h х w, состоящего из символов symb. h, w  и symb - вводит пользователь на отдельных строках. 

```python
h = int(input())
w = int(input())
symb = input()
for _ in range(h):
    print(symb * w)
```
