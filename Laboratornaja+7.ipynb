 > Преподавателю захотелось узнать, кто из учеников присутствовал на всех уроках с начала учебного года. Для каждого урока у него есть список присутствовавших учеников. Фамилии в сп исках не обязательно идут в алфавитном порядке. Определите фамилии учеников, которые присутствовали на всех уроках. 
 
В первой строчке указано число занятий M. Далее идёт M блоков строчек, описывающих списки фамилий. На первой строчке каждого такого блока указано количество фамилий Ni, затем идёт  Ni строчек с фамилиями тех, кто был на уроке i. 

```python
M = int(input("Количество дней"))
if M > 0:
    N1 = int(input("Количество учеников в первый день"))
    present = set(input("Фамилия ученика").strip() for _ in range(N1))
    for _ in range(M - 1):
        Ni = int(input("Количество учеников в следующий день"))
        current = set(input("Фамилия ученика").strip() for _ in range(Ni))
        present.intersection_update(current)
    for name in sorted(present):
        print(name)

# Пример реализации
"""
3
3
Иванов
Петров
Сидоров
2
Иванов
Петров
4
Кузнецов
Иванов
Петров
Сидоров
"""
```

 > ## EX-1-3
 
 >
 
 > Напишите программу, которая кодирует строку цифр в RLE. RLE означает “run-length encoding” — это способ сокращённой записи последовательности чего угодно (в случае этой задачи —  цифр), при котором для подряд идущих группы одинаковых цифр (run) записываются длина этой группы (run length) и сама эта цифра. 

```python
s = input("Строка цифр").strip()
if s:
    res = []
    count = 1
    for i in range(1, len(s)):
        if s[i] == s[i-1]:
            count += 1
        else:
            res.append(f"{count}{s[i-1]}")
            count = 1
    res.append(f"{count}{s[-1]}")
    print(" ".join(res))
else:
    print()

# Пример реализации
# 11122333311
```

 > ## EX-1-4

 > Напишите программу, которая генерирует последовательность целых чисел, строющуюся по следующему правилу: начальный член последовательности равен нулю, каждый следующий член после довательности равен количеству совпадений между уже построенным куском последовательности и им же в обратном порядке. 

```python
N = int(input("Введите число"))
seq = [0]
print(seq[0], end=" ")
for i in range(1, N):
    matches = sum(1 for x, y in zip(seq, reversed(seq)) if x == y)
    seq.append(matches)
    print(seq[-1], end=" ")
print()

# Пример реализации
# 7
```

 > ## EX-1-6
 
 >
 
 > Напишите программу, которая считывает строку текста и производит транслитерацию этой строки с сохранением регистра и пунктуации, т.е. заменить русские буквы на английские. Таблиц у транслитерации можете найти здесь - http://mishka.travel/blog/index/node/id/1073-transliteraciya-v-zagranpasportah-rossiiskoi-federacii/
 
 
 
Букву «ё» транслитерируйте как «e», «й» как «и», а «ъ» и «ь» (и их заглавные версии «Ъ» и «Ь») должны исчезнуть из текста. Строчные буквы заменяются на строчные, заглавные заменяю тся на заглавные. Если заглавная буква превращается при транслитерации в несколько букв, то заглавной должна остаться только первая из них (например, «Ц» → «Tc»).
 
 
 
Все некириллические символы должны остаться на месте. 

```python
def transliterate(s):
    mapping = {
        'а': 'a', 'б': 'b', 'в': 'v', 'г': 'g', 'д': 'd', 'е': 'e', 'ё': 'e',
        'ж': 'zh', 'з': 'z', 'и': 'i', 'й': 'i', 'к': 'k', 'л': 'l', 'м': 'm',
        'н': 'n', 'о': 'o', 'п': 'p', 'р': 'r', 'с': 's', 'т': 't', 'у': 'u',
        'ф': 'f', 'х': 'kh', 'ц': 'ts', 'ч': 'ch', 'ш': 'sh', 'щ': 'shch',
        'ъ': '', 'ы': 'y', 'ь': '', 'э': 'e', 'ю': 'yu', 'я': 'ya'
    }
    res = []
    for char in s:
        if char.lower() in mapping:
            t = mapping[char.lower()]
            if char.isupper() and t:
                res.append(t.capitalize())
            else:
                res.append(t)
        else:
            res.append(char)
    return "".join(res)

print(transliterate(input("Введите строку")))

# Пример реализации
# Ёмае, как же съесть ещё больше
```

 > ## EX-1-7
 
 >
 
 > Дан текст на русском языке. Посчитайте, сколько раз встречается в тексте каждое слово и выведите их  упорядоченными по убыванию количества их попаданий в текст.  
 
Слово — это последовательность русских букв. Кроме слов, в тексте могут присутствовать пробелы и знаки препинания («,», «.», «!», «?», «:» и «;»). Других символов в тексте нет.
 
Слова необходимо сравнивать с точностью до регистра. То есть «Привет» и «привет» — это одно и то же слово, а вот «рыба», «рыбы» и «рыбу» — три разных. 
 
Программа сперва считывает количество строк текста N, а затем сами N-cтрок. 

```python
import re
from collections import Counter

N = int(input())
text = " ".join(input() for _ in range(N))
words = re.findall(r'[а-яА-ЯёЁ]+', text)
words_lower = [w.lower() for w in words]
counts = Counter(words_lower)
for word, count in sorted(counts.items(), key=lambda x: x[1], reverse=True):
    print(word, count)
```

 > Напишите функцию triangle(a, b, c), которая принимает на вход три длины отрезков и определяет, можно ли из этих отрезков составить треугольник. Ваша функция должна печатать «Это  треугольник», если составить треугольник можно, и «Это не треугольник», если нельзя. 

```python
def triangle(a, b, c):
    if a > 0 and b > 0 and c > 0 and a + b > c and a + c > b and b + c > a:
        print("Это треугольник")
    else:
        print("Это не треугольник")


triangle(3,4,5) # Это треугольник
triangle(1,2,3) # Это не треугольник
```

 > Вам необходимо написать функцию golden_ratio(i), которая получает число i как аргумент и печатает i-ое последовательное приближение золотого сечения. i-ое приближения золотого се чения вычисляется как отношение (i+1)-го числа Фибоначчи к i-му.
 
 >  
 
 > Числами Фибоначчи называется последовательность, первые два элемента которой (нумерация с единицы) равны 1, а каждое из последующих чисел равно сумме двух предыдущих. Первые член ы последовательности Фибоначчи - 1, 1, 2, 3, 5, 8, 13, ...
 
 >  
 
 > Таким образом первое приближение золотого сечения — 1/1, второе — 2/1, третье — 3/2, четвертое — 5/3 и т. д. 

```python
def golden_ratio(i):
    if i == 1:
        print(1.0)
        return
    a, b = 1, 1
    for _ in range(2, i + 1):
        a, b = b, a + b
    print(b / a)


golden_ratio(1) # 1 = 1/1
golden_ratio(4) # 1.66(6) = 5/3 
```

 > ## EX-2-4
 
 >
 
 > Правильной скобочной последовательностью называется строка, состоящая только из символов «скобки» (открывающих  "(" и закрывающих ")" ), где каждой закрывающей скобке найдётся соответствующая открывающая. Например, () и (()()) — правильные последовательности, а (()(() или )( — нет.
 
 >  
 
 >Напишите функцию bracket_check(test_string), которая проверяет, является ли поступившая на вход строка правильной скобочной последовательностью. Если да, она должна печатать YES,  в противном случае — NO. Обратите внимание, что пустая строка также является правильной скобочной последовательностью. 

```python
def bracket_check(test_string):
    balance = 0
    for char in test_string:
        if char == '(':
            balance += 1
        elif char == ')':
            balance -= 1
        if balance < 0:
            print("NO")
            return
    print("YES" if balance == 0 else "NO")


bracket_check("((()))") # YES
bracket_check("())") # NO
```

 > Напишите функцию print_long_words(text), которая будет искать в переданной строке слова с четырьмя и более слогами (гласными буквами). Строка передается параметром функции. Слово м следует считать набор букв, ограниченный любыми символами, не являющимися буквами (либо началом/концом строки).
 
 >  
 
 >Найденные слова должны быть выведены на экран в порядке, в котором они встретились. Каждое слово должно идти на своей строке и печататься маленькими буквами.
 
 >  
 
 >В русском языке гласными являются буквы: а, о, э, и, у, ы, е, ё, ю, я. В английском: a, e, i, o, u, y.
 
 >  
 
 >Как и в прочих заданиях этого урока, в вашем решении функция должна быть определена, но не должна вызываться. 

```python
import re

def print_long_words(text):
    vowels = set("аоэиуыеёюяaeiouyАОЭИУЫЕЁЮЯAEIOUY")
    words = re.findall(r'[a-zA-Zа-яА-ЯёЁ]+', text)
    for word in words:
        if sum(1 for c in word if c in vowels) >= 4:
            print(word.lower())

    
print_long_words("Фундаментальная информатика и информационные технологии. Краткие слова")
```

 > ## EX-2-6
 
 >
 
 > Напишите функцию same_by(characteristic, objects), которая проверяет, все ли объекты имеют одинаковое значение некоторой характеристики, и возвращает True, если это так. Если зна чение характеристики для разных объектов отличается – то False. Для пустого набора объектов, функция должна возвращать True. Аргумент characteristic – это функция, которая принимае т объект и вычисляет его характеристику. 

```python
def same_by(characteristic, objects):
    if not objects:
        return True
    return len(set(map(characteristic, objects))) == 1


print(same_by(len, ["cat", "dog", "bog"])) # True
print(same_by(type, [1, 1.0, 2])) # False
```

 > Напишите функцию, которая создает, заполняет и возвращает матрицу заданного размера. При этом (в зависимости от переданных аргументов) она должна вести себя так:
 
 - matrix() — создает матрицу 1×1, в которой единственное число равно нулю.
 
 - matrix(n) — создает матрицу n×n, заполненную нулями.
 
 - matrix(n, m) — создает матрицу из n cтрок и m столбцов, заполненную нулями.
 
 - matrix(n, m, a) — создает матрицу из n строк и m столбцов, в которой каждый элемент равен a. 

```python
def matrix(n=1, m=None, a=0):
    if m is None:
        m = n
    return [[a] * m for _ in range(n)]


print(matrix(2,3)) # [[0,0,0],[0,0,0]]
print(matrix(3,2,-1)) # [[-1,-1],[-1,-1],[-1,-1]]
```
